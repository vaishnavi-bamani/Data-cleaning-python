# 🎯 Objective

Data Types define the kind of values stored in each column of a dataset.

A dataset may contain:

- Numbers
- Text
- Dates
- Boolean Values
- Categories

Sometimes, data is stored in the wrong format.

For example:

| Value | Correct Data Type | Wrong Data Type |
|--------|-------------------|-----------------|
| 25 | Integer | Object |
| 55000 | Float | Object |
| 2024-01-15 | Datetime | Object |
| True | Boolean | Object |

Incorrect data types can lead to:

❌ Calculation Errors

❌ Sorting Problems

❌ Visualization Errors

❌ Machine Learning Issues

In this chapter, we will learn how to inspect, convert, and optimize data types.

> **Rule #1:** Always verify data types before analysis.

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

# 🔢 Convert Data Type using `astype()`

The `astype()` function converts one data type into another.

### Syntax

```python
df["column"] = df["column"].astype(new_dtype)
```

````python
# ==========================================
# Convert Age to Integer
# ==========================================

temp_df = df.copy()

temp_df["age"] = temp_df["age"].astype("Int64")

temp_df.dtypes
````

````python
# ==========================================
# Convert Salary to Float
# ==========================================

temp_df = df.copy()

temp_df["salary"] = temp_df["salary"].astype("float")

temp_df.dtypes
````

````python
# ==========================================
# Convert Employee ID to String
# ==========================================

temp_df = df.copy()

temp_df["employee_id"] = temp_df["employee_id"].astype("string")

temp_df.dtypes
````

````python
# ==========================================
# Convert Department to Category
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].astype("category")

temp_df.dtypes
````

`pd.to_numeric()` converts text into numbers.

### Syntax

```python

```# 🔢 Convert to Numeric using `pd.to_numeric()`

Sometimes numbers are stored as text.

Example:

```
"25000"
"45000"
"N/A"
"Unknown"
```

`pd.to_numeric()` converts text into numbers.

### Syntax

```python
pd.to_numeric(column)
```

````python
# ==========================================
# Convert Salary to Numeric
# ==========================================

temp_df = df.copy()

temp_df["salary"] = pd.to_numeric(
    temp_df["salary"],
    errors="coerce"
)

temp_df.dtypes
````

# ⚠️ Understanding `errors`

The `errors` parameter controls what happens when invalid values are found.

Options:

- `"raise"` (Default)
- `"coerce"`
- `"ignore"`

````python
# ==========================================
# Convert Invalid Values into NaN
# ==========================================

temp_df = df.copy()

temp_df["salary"] = pd.to_numeric(
    temp_df["salary"],
    errors="coerce"
)

temp_df.head()
````

# 📅 Convert to Datetime

Date columns are often stored as text.

Use `pd.to_datetime()` to convert them into proper datetime format.

### Syntax

```python
pd.to_datetime(column)
```

````python
# ==========================================
# Convert Joining Date
# ==========================================

temp_df = df.copy()

temp_df["joining_date"] = pd.to_datetime(
    temp_df["joining_date"],
    errors="coerce"
)

temp_df.dtypes

# ==========================================
# Display Converted Dates
# ==========================================

temp_df["joining_date"].head(10)
````

# 🔄 Convert Multiple Columns

Multiple columns can be converted together.

````python
# ==========================================
# Convert Multiple Columns
# ==========================================

temp_df = df.copy()

temp_df = temp_df.astype({
    "employee_id": "string",
    "department": "category"
})

temp_df.dtypes
````

# 🚀 Automatically Detect Better Data Types

Pandas provides `convert_dtypes()` to automatically choose better data types.

````python
# ==========================================
# Automatic Data Type Conversion
# ==========================================

temp_df = df.copy()

temp_df = temp_df.convert_dtypes()

temp_df.dtypes
````

# 💾 Check Memory Usage

Using proper data types can reduce memory consumption.

# ==========================================

````python
# ==========================================
# Total Memory Usage (MB)
# ==========================================

df.memory_usage(deep=True).sum() / 1024**2
````

# 📝 Best Practices

✅ Always inspect data types before analysis.

✅ Use `astype()` for simple conversions.

✅ Use `pd.to_numeric()` for numeric text.

✅ Use `pd.to_datetime()` for dates.

✅ Convert repetitive text columns to `category`.

✅ Use `convert_dtypes()` to improve data types automatically.

❌ Never assume a column has the correct data type.
