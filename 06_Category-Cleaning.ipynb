# 🎯 Objective

Categorical columns contain a limited number of repeated values.

Examples:

- Department
- Gender
- City
- Country
- Education
- Marital Status

Real-world datasets often contain inconsistent category names due to:

- Typing mistakes
- Different capitalizations
- Extra spaces
- Abbreviations
- Different spellings

These inconsistencies can lead to incorrect analysis and misleading visualizations.

In this chapter, we will learn how to identify, standardize, and clean categorical data.

> **Rule #1:** Every category should have one standard representation.

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

````python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

raw_df = df.copy()
````

# 👀 View Categorical Columns

Let's first inspect the categorical columns.

````python
# ==========================================
# Display Selected Columns
# ==========================================

df[["department","gender","city"]].head(10)
````

# 🔍 Find Unique Values

The `unique()` function displays every unique category present in a column.

### Syntax

```python
Series.unique()
```

````python
# ==========================================
# Unique Departments
# ==========================================

df["department"].unique()

# ==========================================
# Unique Cities
# ==========================================

df["city"].unique()

# ==========================================
# Unique Gender
# ==========================================

df["gender"].unique()
````

# 🔢 Count Unique Categories

Instead of displaying values,

count how many unique categories exist.

### Syntax

```python
Series.nunique()
```

````python
# ==========================================
# Count Unique Departments
# ==========================================

df["department"].nunique()

# ==========================================
# Count Unique Cities
# ==========================================

df["city"].nunique()
````

# 📊 Frequency of Categories

The `value_counts()` function shows how frequently each category appears.

### Syntax

```python
Series.value_counts()
```

````python
# ==========================================
# Department Frequency
# ==========================================

df["department"].value_counts()

# ==========================================
# City Frequency
# ==========================================

df["city"].value_counts()

# ==========================================
# Gender Frequency
# ==========================================

df["gender"].value_counts(dropna=False)
````

# 🔄 Replace Category Values

The `replace()` function replaces incorrect category names.

### Syntax

```python
Series.replace(old,new)
```

````python
# ==========================================
# Replace Single Category
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].replace(
    "HR Dept",
    "HR"
)

temp_df["department"].unique()

# ==========================================
# Replace Multiple Categories
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].replace({

    "Sales Dept":"Sales",

    "sales":"Sales",

    "SALES":"Sales",

    "Hr":"HR",

    "Human Resource":"HR"

})

temp_df["department"].unique()
````

# 🗺️ Map Categories

The `map()` function replaces values using a dictionary.

Useful when converting short codes into meaningful labels.

### Syntax

```python
Series.map(dictionary)
```

````python
# ==========================================
# Gender Mapping
# ==========================================

temp_df = df.copy()

gender_map = {

    "M":"Male",

    "F":"Female"

}

temp_df["gender"] = temp_df["gender"].map(gender_map)

temp_df["gender"].head()

# ==========================================
# Rating Mapping
# ==========================================

temp_df = df.copy()

rating_map = {

    1:"Poor",

    2:"Average",

    3:"Good",

    4:"Very Good",

    5:"Excellent"

}

temp_df["performance_score"] = temp_df["performance_score"].map(rating_map)

temp_df.head()
````

# 📋 Standardize Categories

Most category cleaning follows the same workflow:

- Remove spaces
- Convert to Title Case
- Replace incorrect spellings

````python
# ==========================================
# Standardize Department
# ==========================================

temp_df = df.copy()

temp_df["department"] = (

    temp_df["department"]

    .str.strip()

    .str.title()

    .replace({

        "Sales Dept":"Sales",

        "Hr":"HR",

        "Human Resource":"HR"

    })

)

temp_df["department"].unique()

# ==========================================
# Standardize City
# ==========================================

temp_df = df.copy()

temp_df["city"] = (

    temp_df["city"]

    .str.strip()

    .str.title()

)

temp_df["city"].unique()
````

# 🚫 Detect Rare Categories

Sometimes categories occur only once due to typing mistakes.

These should be investigated.

````python
# ==========================================
# Categories Appearing Only Once
# ==========================================

department_counts = df["department"].value_counts()

department_counts[department_counts == 1]
````

# 📋 Category Cleaning Cheat Sheet

| Function | Purpose |
|----------|---------|
| `unique()` | Show unique values |
| `nunique()` | Count unique values |
| `value_counts()` | Frequency of categories |
| `replace()` | Replace categories |
| `map()` | Map categories |
| `str.strip()` | Remove spaces |
| `str.title()` | Standardize capitalization |
