# 🎯 Objective

Date and Time columns are among the most important columns in any dataset.

Unfortunately, they are often stored as plain text (Object Data Type).

Examples:

```
2025-01-15
15/01/2025
Jan 15, 2025
01-15-2025
```

Although these values represent dates, Pandas cannot perform date operations until they are converted into the proper datetime format.

After converting dates, we can:

✅ Extract Year

✅ Extract Month

✅ Extract Day

✅ Calculate Date Difference

✅ Sort Chronologically

✅ Filter Date Ranges

✅ Detect Invalid Dates

> **Rule #1:** Always convert date columns to datetime before analysis.

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

````python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

raw_df = df.copy()
````

# 🔍 Check Current Data Type

Before converting,

let's inspect the current datatype.

````python
# ==========================================
# Current Data Type
# ==========================================

df["joining_date"].dtype

# ==========================================
# Dataset Information
# ==========================================

df.info()
````

# 🔄 Convert to Datetime

Pandas provides the `pd.to_datetime()` function.

### Syntax

```python
pd.to_datetime(column)
```

````python
# ==========================================
# Convert Joining Date
# ==========================================

temp_df = df.copy()

temp_df["joining_date"] = pd.to_datetime(
    temp_df["joining_date"],
    errors="coerce"
)

temp_df["joining_date"].head()
````

# ⚠️ Handling Invalid Dates

Sometimes the dataset contains invalid values.

Example

```
31/02/2025

abc

Not Available
```

Using

```python
errors="coerce"
```

invalid values automatically become **NaT (Not a Time)**.

````python
# ==========================================
# Invalid Dates become NaT
# ==========================================

temp_df = df.copy()

temp_df["joining_date"] = pd.to_datetime(
    temp_df["joining_date"],
    errors="coerce"
)

temp_df[temp_df["joining_date"].isna()]
````

# 📅 Extract Year

````python
# ==========================================
# Extract Year
# ==========================================

temp_df = df.copy()

temp_df["joining_date"] = pd.to_datetime(
    temp_df["joining_date"],
    errors="coerce"
)

temp_df["joining_year"] = temp_df["joining_date"].dt.year

temp_df.head()
````

# 📅 Extract Month

````python
# ==========================================
# Extract Month
# ==========================================

temp_df["joining_month"] = temp_df["joining_date"].dt.month

temp_df.head()
````

# 📅 Extract Month Name

````python
# ==========================================
# Month Name
# ==========================================

temp_df["month_name"] = temp_df["joining_date"].dt.month_name()

temp_df.head()
````

# 📅 Extract Day

````python
# ==========================================
# Day
# ==========================================

temp_df["joining_day"] = temp_df["joining_date"].dt.day

temp_df.head()
````

# 📅 Day Name

````python
# ==========================================
# Day Name
# ==========================================

temp_df["day_name"] = temp_df["joining_date"].dt.day_name()

temp_df.head()
````

# 📅 Extract Weekday Number

````python
# ==========================================
# Weekday Number
# ==========================================

temp_df["weekday"] = temp_df["joining_date"].dt.weekday

temp_df.head()
````

# 📅 Extract Quarter

````python
# ==========================================
# Quarter
# ==========================================

temp_df["quarter"] = temp_df["joining_date"].dt.quarter

temp_df.head()
````

# 📅 Extract Week Number

````python
# ==========================================
# ISO Week Number
# ==========================================

temp_df["week"] = temp_df["joining_date"].dt.isocalendar().week

temp_df.head()
````

# 🕒 Extract Hour, Minute & Second

Useful when timestamps are available.

````python
# ==========================================
# Hour
# ==========================================

temp_df["hour"] = temp_df["joining_date"].dt.hour

temp_df.head()

# ==========================================
# Minute
# ==========================================

temp_df["minute"] = temp_df["joining_date"].dt.minute

temp_df.head()

# ==========================================
# Second
# ==========================================

temp_df["second"] = temp_df["joining_date"].dt.second

temp_df.head()
````

# 📆 Date Difference

Calculate how many days have passed since joining.

````python
# ==========================================
# Days Since Joining
# ==========================================

today = pd.Timestamp.today()

temp_df["days_since_joining"] = (
    today - temp_df["joining_date"]
).dt.days

temp_df.head()
````

# 📅 Format Dates

Convert datetime into a custom format.

### Syntax

```python
dt.strftime()
```

````python
# ==========================================
# Format Date
# ==========================================

temp_df["joining_date"].dt.strftime("%d-%m-%Y")

# ==========================================
# Month-Year Format
# ==========================================

temp_df["joining_date"].dt.strftime("%B %Y")
````

# 🔍 Filter Dates

````python
# ==========================================
# Employees Joined After 2022
# ==========================================

temp_df[
    temp_df["joining_date"] >= "2022-01-01"
]

# ==========================================
# Employees Joined in 2023
# ==========================================

temp_df[
    temp_df["joining_date"].dt.year == 2023
]
````
