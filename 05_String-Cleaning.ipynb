# 🔠 Convert Text to Lowercase

Sometimes the same category appears in different letter cases.

Example

Sales

sales

SALES

These should all become

sales

### Syntax

```python
Series.str.lower()
```

````python
# ==========================================
# Convert Name to Lowercase
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.lower()

temp_df["full_name"].head(10)

# ==========================================
# Convert Department to Lowercase
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].str.lower()

temp_df["department"].head(10)
````

# 🔠 Convert Text to Uppercase

Converts every letter into uppercase.

### Syntax

```python
Series.str.upper()
```

````python
# ==========================================
# Convert Name to Uppercase
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.upper()

temp_df["full_name"].head(10)

# ==========================================
# Convert City to Uppercase
# ==========================================

temp_df = df.copy()

temp_df["city"] = temp_df["city"].str.upper()

temp_df["city"].head(10)
````

# ✨ Convert Text to Title Case

Title Case capitalizes the first letter of every word.

Example

john doe

↓

John Doe

### Syntax

```python
Series.str.title()
```

````python
# ==========================================
# Convert Name to Title Case
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.title()

temp_df["full_name"].head(10)

# ==========================================
# Convert Department to Title Case
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].str.title()

temp_df["department"].head(10)
````

# 🔤 Capitalize First Letter

Series.str.capitalize()

````python
# ==========================================
# Capitalize Name
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.capitalize()

temp_df["full_name"].head(10)
````

# 🔄 Swap Letter Cases

Uppercase becomes lowercase.

### Syntax

```python
Series.str.swapcase()
```

````python
# ==========================================
# Swap Letter Cases
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].str.swapcase()

temp_df["department"].head(10)
````

# ✂️ Removing Unwanted Spaces

---

# 🎯 Objective

Extra spaces are one of the most common problems in text data.

Examples:

```
" John Doe "
"  Sales"
"Pune   "
"  HR  "
```

Although these values may look identical,

Python considers them different because of the extra spaces.

Pandas provides powerful string functions to remove unwanted spaces.

> **Rule #2:** Always remove unnecessary spaces before cleaning text data.

````python
# ==========================================
# Display Original Data
# ==========================================

df["full_name"].head(10)
````

# ✂️ Remove Leading and Trailing Spaces

The `str.strip()` function removes spaces from both the beginning and the end of a string.

### Syntax

```python
Series.str.strip()
```

````python
# ==========================================
# Remove Leading & Trailing Spaces
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.strip()

temp_df["full_name"].head(10)

# ==========================================
# Remove Spaces from Department
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].str.strip()

temp_df["department"].head(10)
````

# ⬅️ Remove Leading Spaces

Sometimes we only want to remove spaces from the beginning.

### Syntax

```python
Series.str.lstrip()
```

````python
# ==========================================
# Remove Leading Spaces
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.lstrip()

temp_df["full_name"].head(10)
````

# ➡️ Remove Trailing Spaces

Sometimes we only want to remove spaces from the end.

### Syntax

```python
Series.str.rstrip()
```

````python
# ==========================================
# Remove Trailing Spaces
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.rstrip()

temp_df["full_name"].head(10)
````

# 🔄 Replace Characters

The `str.replace()` function replaces one value with another.

It is one of the most frequently used functions during data cleaning.

### Syntax

```python
Series.str.replace(old, new)
```

````python
# ==========================================
# Replace "-" with Space
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.replace("-", " ")

temp_df["full_name"].head(10)

# ==========================================
# Replace "_" with Space
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].str.replace("_", " ")

temp_df["department"].head(10)

# ==========================================
# Remove "." from Names
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.replace(".", "", regex=False)

temp_df["full_name"].head(10)

# ==========================================
# Remove "," from City
# ==========================================

temp_df = df.copy()

temp_df["city"] = temp_df["city"].str.replace(",", "", regex=False)

temp_df["city"].head(10)
````

The `str.endswith()` function checks whether a string ends with specific text.

### Syntax

```python
Series.str.endswith(".com")
```

````python
# ==========================================
# Gmail Users
# ==========================================

df[df["email"].str.endswith("@gmail.com", na=False)]
````

# ✂️ Splitting, Joining & Extracting Text

---

# 🎯 Objective

Sometimes a single text column contains multiple pieces of information.

Examples:

```
John Doe
john@gmail.com
EMP-1001
Pune, Maharashtra
```

Instead of keeping everything together,

we can split, extract, combine, and manipulate text using Pandas String Functions.

These functions are extremely useful while cleaning real-world datasets.

> **Rule #3:** Never manually split text. Use Pandas string functions.

````python
# ==========================================
# Display Sample Columns
# ==========================================

df[["full_name", "email"]].head()
````

# ✂️ Split Text using `str.split()`

The `str.split()` function splits a string into multiple parts.

### Syntax

```python
Series.str.split(separator)
```

````python
# ==========================================
# Split Full Name
# ==========================================

df["full_name"].str.split(" ")

# ==========================================
# Split Email
# ==========================================

df["email"].str.split("@")

# ==========================================
# Split Name into Multiple Columns
# ==========================================

temp_df = df.copy()

temp_df[["first_name", "last_name"]] = temp_df["full_name"].str.split(
    " ",
    n=1,
    expand=True
)

temp_df.head()
````

# 📍 Access Split Values

After splitting,

individual parts can be accessed using indexing.

````python
# ==========================================
# First Name
# ==========================================

df["full_name"].str.split().str[0]

# ==========================================
# Last Name
# ==========================================

df["full_name"].str.split().str[-1]
````

# 🔗 Join Strings using `str.cat()`

The `str.cat()` function joins multiple strings together.

### Syntax

```python
Series.str.cat(other)
```

````python
# ==========================================
# Combine Name and Department
# ==========================================

df["full_name"].str.cat(
    df["department"],
    sep=" - "
)

# ==========================================
# Combine Name and City
# ==========================================

df["full_name"].str.cat(
    df["city"],
    sep=", "
)
````

# ✂️ Extract Text using `str.extract()`

Sometimes we only need a specific part of a string.

`str.extract()` uses Regular Expressions (Regex).

### Syntax

```python
Series.str.extract(pattern)
```

````python
# ==========================================
# Extract Email Username
# ==========================================

df["email"].str.extract(r"(^.*)@")

# ==========================================
# Extract Numbers from Employee ID
# ==========================================

df["employee_id"].astype(str).str.extract(r"(\d+)")
````

# ✂️ Slice Text using `str.slice()`

Extract a specific portion of a string.

### Syntax

```python
Series.str.slice(start, stop)
```

````python
# ==========================================
# First Three Characters
# ==========================================

df["full_name"].str.slice(0,3)

# ==========================================
# Last Four Characters
# ==========================================

df["full_name"].str.slice(-4)

# ==========================================
# First Five Characters of Email
# ==========================================

df["email"].str.slice(0,5)
````

# 🔁 Repeat Text using `str.repeat()`

Repeats each string multiple times.

### Syntax

```python
Series.str.repeat(n)
```

````python
# ==========================================
# Repeat Name Twice
# ==========================================

df["full_name"].str.repeat(2)
````

# 0️⃣ Pad Numbers using `str.zfill()`

Adds leading zeros until a fixed width is reached.

Useful for Employee IDs and Invoice Numbers.

### Syntax

```python
Series.str.zfill(width)
```

````python
# ==========================================
# Employee ID with Leading Zeros
# ==========================================

df["employee_id"].astype(str).str.zfill(6)
````

# 📏 Pad Text using `str.pad()`

Pads a string with characters on the left, right, or both sides.

### Syntax

```python
Series.str.pad(width)
```

````python
# ==========================================
# Left Pad
# ==========================================

df["employee_id"].astype(str).str.pad(
    width=8,
    side="left",
    fillchar="0"
)

# ==========================================
# Right Pad
# ==========================================

df["employee_id"].astype(str).str.pad(
    width=8,
    side="right",
    fillchar="-"
)

# ==========================================
# Both Side Padding
# ==========================================

df["employee_id"].astype(str).str.pad(
    width=10,
    side="both",
    fillchar="*"
)
````

# 🔥 Regular Expressions (Regex) for String Cleaning

---

# 🎯 Objective

Real-world text data is often messy.

Examples:

```
John@123
Rahul!!!
Pune#$
98765-43210
john.doe@gmail.com
```

Regular Expressions (Regex) help us search, extract, validate, and replace text using patterns instead of exact words.

Regex is one of the most powerful tools in Data Cleaning.

> **Rule #4:** Learn Regex patterns—they can save hours of manual cleaning.

# 📖 Common Regex Patterns

| Pattern | Meaning | Example |
|----------|----------|---------|
| `\d` | Any digit | 0-9 |
| `\D` | Non-digit | A, @, # |
| `\w` | Letter, Number, _ | abc123 |
| `\W` | Special Character | @#$ |
| `\s` | White Space | Space, Tab |
| `\S` | Non White Space | A, 1 |
| `^` | Starts With | ^A |
| `$` | Ends With | com$ |
| `+` | One or More | \d+ |
| `*` | Zero or More | A* |
| `.` | Any Character | a.c |

# 🔎 Find Names Containing Numbers

Sometimes names contain invalid characters like:

```
John123
Rahul45
Ankit007
```

Let's find them.

````python
# ==========================================
# Names Containing Numbers
# ==========================================

df[df["full_name"].str.contains(r"\d", na=False)]
````

# 🔎 Find Emails Ending with Gmail

````python
# ==========================================
# Gmail Users
# ==========================================

df[df["email"].str.contains(r"@gmail\.com$", na=False)]
````

# 🔎 Find Phone Numbers Having Only Digits

````python
# ==========================================
# Valid Phone Numbers
# ==========================================

df[df["phone"].astype(str).str.match(r"^\d+$", na=False)]
````

# 🧹 Remove Numbers from Names

````python
# ==========================================
# Remove Numbers
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.replace(
    r"\d+",
    "",
    regex=True
)

temp_df["full_name"].head()
````

# 🧹 Remove Special Characters

````python
# ==========================================
# Remove Special Characters
# ==========================================

temp_df = df.copy()

temp_df["full_name"] = temp_df["full_name"].str.replace(
    r"[^A-Za-z ]",
    "",
    regex=True
)

temp_df["full_name"].head()
````

# 🧹 Keep Only Numbers

````python
# ==========================================
# Keep Digits Only
# ==========================================

temp_df = df.copy()

temp_df["phone"] = temp_df["phone"].astype(str).str.replace(
    r"\D",
    "",
    regex=True
)

temp_df["phone"].head()
````

# ✉️ Validate Email Addresses

````python
# ==========================================
# Valid Emails
# ==========================================

valid_email = df["email"].str.match(
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$",
    na=False
)

df[valid_email]

# ==========================================
# Invalid Emails
# ==========================================

invalid_email = ~df["email"].str.match(
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$",
    na=False
)

df[invalid_email]
````

# 📞 Clean Phone Numbers

Convert

```
98765-43210
98765 43210
+91 9876543210
(98765)43210
```

into

```
9876543210
```

````python
# ==========================================
# Clean Phone Numbers
# ==========================================

temp_df = df.copy()

temp_df["phone"] = temp_df["phone"].astype(str).str.replace(
    r"\D",
    "",
    regex=True
)

temp_df["phone"].head()
````

# 👤 Clean Employee Names

Convert

```
john123
JOHN##
John___
```

into

```
John
```

````python
# ==========================================
# Clean Employee Names
# ==========================================

temp_df = df.copy()

temp_df["name"] = (
    temp_df["full_name"]
    .str.replace(r"[^A-Za-z ]", "", regex=True)
    .str.strip()
    .str.title()
)

temp_df["full_name"].head()
````

# 🏙️ Standardize City Names

````python
# ==========================================
# Clean City Names
# ==========================================

temp_df = df.copy()

temp_df["city"] = (
    temp_df["city"]
    .str.strip()
    .str.title()
)

temp_df["city"].head()
````

# 🏢 Standardize Department Names

````python
# ==========================================
# Clean Department Names
# ==========================================

temp_df = df.copy()

temp_df["department"] = (
    temp_df["department"]
    .str.strip()
    .str.title()
)

temp_df["department"].head()
````
