# 🔍 Dataset Audit

---

## 🎯 Objective

Before cleaning any dataset, we need to inspect its quality.

A Data Audit helps us answer questions like:

- How many missing values are there?
- Are there duplicate records?
- What are the data types?
- Which columns have the most unique values?
- Which columns need immediate attention?

Think of this step as a **Health Checkup** for your dataset.

> **Rule #2:** Always audit your data before cleaning it.

```python
# ==========================================
# 📊 Display Complete Dataset Information
# ==========================================

df.info()

# ==========================================
# 📋 Check Data Types
# ==========================================

df.dtypes

# ==========================================
# 📏 Dataset Dimensions
# ==========================================

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

# ==========================================
# 🏷️ Display All Column Names
# ==========================================

df.columns.tolist()
```

# ❓ Are There Any Missing Values?

# Missing values are one of the most common problems in real-world datasets.
# Let's find out how many missing values each column contains.

```python
# ==========================================
# 🔍 Missing Values Count
# ==========================================

df.isnull().sum()

# ==========================================
# 📊 Missing Values Percentage
# ==========================================

(df.isnull().sum() / len(df) * 100).round(2)

# ==========================================
# 📋 Missing Value Report
# ==========================================

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df)*100).round(2)
})

missing_report.sort_values(
    by="Missing Values",
    ascending=False
)
```

# 🔁 Duplicate Records

Duplicate records increase dataset size and may lead to incorrect analysis.

Let's check if our dataset contains duplicate rows.

```python
# ==========================================
# 🔁 Total Duplicate Rows
# ==========================================

df.duplicated().sum()

# ==========================================
# 👀 Display Duplicate Rows
# ==========================================

df[df.duplicated()]
```

# 🎯 Unique Values

Knowing the number of unique values helps us identify:

- Categorical columns
- High-cardinality columns
- Possible inconsistencies

```python
# ==========================================
# 🔢 Unique Values Per Column
# ==========================================

df.nunique()

# ==========================================
# 📊 Unique Value Report
# ==========================================

unique_report = pd.DataFrame({
    "Unique Values": df.nunique(),
    "Data Type": df.dtypes
})

unique_report
```

# 📈 Statistical Summary

Let's inspect the numerical columns.

This helps us identify:

- Minimum values
- Maximum values
- Mean
- Standard Deviation
- Possible Outliers

```python
# ==========================================
# 📊 Numerical Summary
# ==========================================

df.describe()

# ==========================================
# 📊 Complete Summary
# ==========================================

df.describe(include="all")
```

# 📋 Data Quality Report

Let's combine everything into a single report.

This report provides a quick overview of the dataset's quality.

```python
# ==========================================
# 📑 Data Quality Report
# ==========================================

quality_report = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().sum()/len(df)*100).round(2),
    "Unique Values": df.nunique(),
    "Duplicate Values": [df[col].duplicated().sum() for col in df.columns]
})

quality_report
```
