# 🎯 Objective

After cleaning and validating the dataset, the final step is to save it in the appropriate file format.

Different formats are used for different purposes:

- CSV → Data Analysis
- Excel → Business Reports
- JSON → APIs & Web Applications
- Parquet → Big Data & Machine Learning
- Pickle → Python Objects

Choosing the correct format improves storage efficiency, compatibility, and performance.

> **Rule #1:** Always validate your data before exporting it.

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

````python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

# Create Backup Copy
raw_df = df.copy()
````

# 📁 Save Dataset as CSV

CSV (Comma Separated Values) is the most widely used file format for storing tabular data.

### Syntax

```python
df.to_csv("filename.csv")
```

````python
# ==========================================
# Save as CSV
# ==========================================

df.to_csv(
    "dataset/clean_employee_data.csv",
    index=False
)
````

# 📊 Save Dataset as Excel

Excel files are commonly used for sharing reports with business users.

### Syntax

```python
df.to_excel("filename.xlsx")
```

````python
# ==========================================
# Save as Excel
# ==========================================

df.to_excel(
    "dataset/clean_employee_data.xlsx",
    index=False
)
````

# 🌐 Save Dataset as JSON

JSON is commonly used in APIs and web applications.

### Syntax

```python
df.to_json("filename.json")
```

````python
# ==========================================
# Save as JSON
# ==========================================

df.to_json(
    "clean_employee_data.json",
    orient="records",
    indent=4
)
````

# 📄 Save Without Index

The `index=False` parameter prevents Pandas from saving row numbers.

Always use it unless the index has meaningful information.

````python
# ==========================================
# Save Without Index
# ==========================================

df.to_csv(
    "employee_no_index.csv",
    index=False
)
````

# 🔍 Verify Exported File

Always reload the exported file to ensure it has been saved correctly.

````python
# ==========================================
# Reload CSV
# ==========================================

check_df = pd.read_csv(
    "dataset/clean_employee_data.csv"
)

check_df.head()
````

# 📊 Compare Original & Exported Dataset

````python
# ==========================================
# Compare Shapes
# ==========================================

print("Original Shape :", df.shape)

print("Reloaded Shape :", check_df.shape)

# ==========================================
# Compare Column Names
# ==========================================

print(df.columns)

print(check_df.columns)
````

# 📋 Export Cheat Sheet

| Function | File Format |
|----------|-------------|
| `to_csv()` | CSV |
| `to_excel()` | Excel |
| `to_json()` | JSON |
| `to_parquet()` | Parquet |
| `to_pickle()` | Pickle |

# 💡 Best Practices

✅ Validate data before exporting.

✅ Use `index=False` unless the index is important.

✅ Choose the correct file format based on the use case.

✅ Verify exported files by loading them again.

✅ Keep raw and cleaned datasets separate.

✅ Use meaningful file names.

Example:

```
employee_raw.csv

employee_clean.csv

employee_final.csv
```
