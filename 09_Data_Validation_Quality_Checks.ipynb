# ✅ Chapter 09 - Data Validation & Quality Checks

# 🎯 Objective

After cleaning a dataset, we should never assume it is perfect.

Before exporting the final dataset, we must validate its quality.

Data Validation helps answer questions like:

- Are there still missing values?
- Are there duplicate records?
- Are data types correct?
- Are values within expected ranges?
- Are email addresses valid?
- Are phone numbers valid?
- Are categories standardized?

Professional Data Analysts always perform a final validation before delivering a dataset.

> **Rule #1:** Never save a dataset without validating it.

```python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
```

```python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

raw_df = df.copy()
```

# 🔍 Check Dataset Shape

Always verify the number of rows and columns.

```python
# ==========================================
# Dataset Shape
# ==========================================

df.shape
```

# 📋 Dataset Information

Check the final data types and non-null values.

```python
# ==========================================
# Dataset Information
# ==========================================

df.info()
```

# ❓ Check Missing Values

```python
# ==========================================
# Missing Values
# ==========================================

df.isnull().sum()

# ==========================================
# Missing Value Percentage
# ==========================================

(df.isnull().sum() / len(df)) * 100
```

# 👥 Check Duplicate Records

```python
# ==========================================
# Duplicate Records
# ==========================================

df.duplicated().sum()

# ==========================================
# Display Duplicate Rows
# ==========================================

df[df.duplicated()]
```

# 🔄 Validate Data Types

```python
# ==========================================
# Data Types
# ==========================================

df.dtypes
```

# 💰 Validate Salary

Salary should always be positive.

```python
# Convert Salary to Numeric
df["salary"] = pd.to_numeric(
    df["salary"],
    errors="coerce"
)

# Now check invalid salaries
df[df["salary"] <= 0]
```

# 🎂 Validate Age

Age should be between 18 and 65.

```python
# ==========================================
# Convert Columns to Numeric
# ==========================================

df["age"] = pd.to_numeric(df["age"], errors="coerce")




# ==========================================
# Invalid Age
# ==========================================

df[
    (df["age"] < 18) |
    (df["age"] > 65)
]
```

# 📧 Validate Email Addresses

```python
# ==========================================
# Invalid Emails
# ==========================================

invalid_email = ~df["email"].str.match(
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$",
    na=False
)

df[invalid_email]
```

# 📞 Validate Phone Numbers

```python
# ==========================================
# Invalid Phone Numbers
# ==========================================

invalid_phone = ~df["phone"].astype(str).str.match(
    r"^\d{10}$",
    na=False
)

df[invalid_phone]
```

# 🏷️ Validate Categories

Check if category values are standardized.

```python
# ==========================================
# Department Categories
# ==========================================

df["department"].value_counts()

# ==========================================
# Gender Categories
# ==========================================

df["gender"].value_counts()

# ==========================================
# City Categories
# ==========================================

df["city"].value_counts()
```

# 📅 Validate Dates

Check for invalid or missing dates.

```python
# ==========================================
# Invalid Joining Dates
# ==========================================

df[df["joining_date"].isna()]
```

# 📊 Statistical Validation

Review summary statistics to identify suspicious values.

```python
# ==========================================
# Numerical Summary
# ==========================================

df.describe()
```

# 🔍 Check Unique Values

```python
# ==========================================
# Unique Value Count
# ==========================================

df.nunique()
```

# 📋 Generate Validation Report

```python
# ==========================================
# Validation Report
# ==========================================

validation_report = pd.DataFrame({

    "Missing Values": df.isnull().sum(),

    "Data Type": df.dtypes,

    "Unique Values": df.nunique()

})

validation_report
```

# ✅ Final Validation Checklist

Before saving your dataset, verify:

- ✔ No unnecessary missing values
- ✔ No duplicate rows
- ✔ Correct data types
- ✔ Valid email addresses
- ✔ Valid phone numbers
- ✔ Correct age range
- ✔ Positive salary values
- ✔ Standardized categories
- ✔ Valid dates
