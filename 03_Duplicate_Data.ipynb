# 🎯 Objective

Duplicate records are one of the most common data quality issues.

A duplicate record means the **same information appears more than once** in a dataset.

Duplicates can occur due to:

- Manual data entry
- Multiple data imports
- System synchronization errors
- Database migration
- Human mistakes

Duplicate data can lead to:

❌ Incorrect Analysis

❌ Incorrect Statistics

❌ Wrong Machine Learning Predictions

❌ Increased Storage

In this chapter, we will learn every important function used to detect, inspect, and remove duplicate records.

> **Rule #1:** Always investigate duplicates before removing them.

# 📚 Import Libraries

import pandas as pd

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

````python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

# Create Backup Copy
raw_df = df.copy()
````

# 🔍 Check Duplicate Rows

The first step is to determine whether duplicate records exist in the dataset.

Pandas provides the `duplicated()` function for this purpose.

### Syntax

```python
df.duplicated()
```

````python
# ==========================================
# 🔍 Check Duplicate Rows
# ==========================================

df.duplicated()
````

# 📊 Count Duplicate Rows

Instead of displaying True and False,

let's count the total number of duplicate rows.

````python
# ==========================================
# 📊 Count Duplicate Rows
# ==========================================

df.duplicated().sum()
````

# 👀 Display Duplicate Rows

Let's inspect the duplicate records before removing them.

````python
# ==========================================
# 👀 Display Duplicate Rows
# ==========================================

df[df.duplicated()]
````

# 🔍 Display All Duplicate Records

The default behavior keeps the first occurrence.

Using `keep=False` displays every duplicate record.

````python
# ==========================================
# 👀 Display Every Duplicate Row
# ==========================================

df[df.duplicated(keep=False)]
````

# 🎯 Check Duplicate Values in a Specific Column

Sometimes the entire row is unique,

but one important column contains duplicates.

Example:

Employee ID

Email

Phone Number

````python
# ==========================================
# Duplicate Employee IDs
# ==========================================

df["employee_id"].duplicated().sum()

# ==========================================
# Duplicate Email Addresses
# ==========================================

df["email"].duplicated().sum()

# ==========================================
# Duplicate Phone Numbers
# ==========================================

df["phone"].duplicated().sum()
````

# 👀 Display Duplicate Employee IDs

````python
# ==========================================
# Display Duplicate Employee IDs
# ==========================================

df[df["employee_id"].duplicated(keep=False)]
````

# 🗑️ Remove Duplicate Rows

Once duplicates have been verified,

they can be removed using `drop_duplicates()`.

### Syntax

```python
df.drop_duplicates()
```

````python
# ==========================================
# Remove Duplicate Rows
# ==========================================

df.drop_duplicates()
````

````python
# ==========================================
# Save Result into New DataFrame
# ==========================================

clean_df = df.drop_duplicates()

clean_df
````

# 📏 Compare Dataset Size

Always compare the dataset before and after removing duplicates.

````python
# ==========================================
# Original Shape
# ==========================================

df.shape

# ==========================================
# Shape After Removing Duplicates
# ==========================================

clean_df.shape
````

# 🎯 Remove Duplicates Based on Selected Columns

Sometimes only specific columns should be checked.

Example:

Employee ID

Email

````python
# ==========================================
# Remove Duplicate Employee IDs
# ==========================================

df.drop_duplicates(subset=["employee_id"])

# ==========================================
# Remove Duplicate Emails
# ==========================================

df.drop_duplicates(subset=["email"])

# ==========================================
# Remove Duplicate Employee ID + Email
# ==========================================

df.drop_duplicates(
    subset=["employee_id", "email"]
)
````

# 🔄 Understanding the `keep` Parameter

The `keep` parameter decides which duplicate record should be retained.

Options:

- `"first"` (Default)
- `"last"`
- `False`

````python
# ==========================================
# Keep First Duplicate
# ==========================================

df.drop_duplicates(
    keep="first"
)

# ==========================================
# Keep Last Duplicate
# ==========================================

df.drop_duplicates(
    keep="last"
)

# ==========================================
# Remove Every Duplicate
# ==========================================

df.drop_duplicates(
    keep=False
)
````

# 🔢 Reset Index

After removing duplicates,

the index may become discontinuous.

Reset the index for a clean DataFrame.

````python
# ==========================================
# Reset Index
# ==========================================

clean_df = df.drop_duplicates()

clean_df.reset_index(drop=True)
````

# 💾 Permanently Remove Duplicate Rows

Use `inplace=True` to modify the original DataFrame.

⚠️ Use carefully because the changes cannot be undone.

````python
# ==========================================
# Permanently Remove Duplicate Rows
# ==========================================

temp_df = df.copy()

temp_df.drop_duplicates(
    inplace=True
)

temp_df
````

# 📋 Verify the Result

After cleaning,

always verify that duplicates have been removed.

````python
# ==========================================
# Verify Duplicate Count
# ==========================================

temp_df.duplicated().sum()
````

# 📝 Best Practices

✅ Always inspect duplicates before removing them.

✅ Check duplicate rows and duplicate IDs separately.

✅ Keep a backup of the original dataset.

✅ Reset the index after removing duplicates.

❌ Never remove duplicates without understanding why they exist.

❌ Some duplicates may represent valid repeated transactions.
