# 🕳️ Chapter 02 - Missing Values

---

# 🎯 Objective

Missing values are one of the most common problems in real-world datasets.

A missing value simply means that some information is unavailable or was not recorded.

Examples:

- Customer forgot to enter their phone number.
- Salary is not available.
- Email address is missing.
- Date of joining is unknown.

If we don't handle missing values properly, they can lead to:

❌ Incorrect Analysis

❌ Wrong Machine Learning Predictions

❌ Runtime Errors

❌ Poor Data Quality

In this chapter, we will learn every important technique used to identify and handle missing values.

> **Rule #1:** Never ignore missing values.

````python
# ==========================================
# 📚 Import Libraries
# ==========================================

import pandas as pd
import numpy as np
````

````python
# ==========================================
# 📂 Load Dataset
# ==========================================

df = pd.read_csv("dataset/employee_dirty_data.csv")

# Backup Copy
raw_df = df.copy()
````

# 🔍 Detect Missing Values

Before fixing missing values, we first need to identify them.

Pandas provides several useful functions for this purpose.

````python
# ==========================================
# 🔍 Check Missing Values
# ==========================================

df.isnull()

# ==========================================
# 🔍 Check Missing Values (Alternative)
# ==========================================

df.isna()
````

# 📌 What is the Difference?

Both functions perform exactly the same task.

```python
df.isnull()
```

and

```python
df.isna()
```

produce identical results.

You can use whichever you prefer.

````python
# ==========================================
# 📊 Count Missing Values
# ==========================================

df.isnull().sum()

# ==========================================
# 📊 Total Missing Values
# ==========================================

df.isnull().sum().sum()

# ==========================================
# 📈 Missing Value Percentage
# ==========================================

(df.isnull().sum()/len(df)*100).round(2)

# ==========================================
# 📋 Missing Value Report
# ==========================================

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df)*100).round(2)
})

missing_report.sort_values(
    by="Missing Values",
    ascending=False
)

````

# 👀 Display Only Columns That Have Missing Values

let's display only the columns that actually contain missing values.

````python
# ==========================================
# 📋 Columns With Missing Values
# ==========================================

df.isnull().sum()[df.isnull().sum() > 0]
````

# ❌ Display Rows That Contain Missing Values

Sometimes we need to inspect the actual rows containing missing values.

````python
# ==========================================
# 🔍 Rows Containing Missing Values
# ==========================================

df[df.isnull().any(axis=1)]
````

# ✅ Display Rows Without Missing Values

This is useful when we want to analyze only complete records.

````python
# ==========================================
# ✅ Rows Without Missing Values
# ==========================================

df[df.notnull().all(axis=1)]
````

# 🗑️ Removing Missing Values using `dropna()`

---

# 🎯 Objective

Sometimes a dataset contains rows or columns with missing values that are no longer useful.

Pandas provides the `dropna()` function to remove missing values from a dataset.

It can remove:

✅ Rows

✅ Columns

✅ Specific missing values

✅ Rows based on a threshold

> **Rule #2:** Never use `dropna()` blindly. Always understand how much data you are removing.

````python
# ==========================================
# 📋 Display Missing Values
# ==========================================

df.isnull().sum()
````

# 🚀 Remove Rows with Missing Values

By default, `dropna()` removes every row that contains at least one missing value.

### Syntax

```python
df.dropna()
```

````python
# ==========================================
# 🗑️ Remove Rows with Missing Values
# ==========================================

df.dropna()
````

````python
# ==========================================
# 🗑️ Save Result into New DataFrame
# ==========================================

clean_df = df.dropna()

clean_df
````

# 📊 Compare Dataset Size

Always compare the dataset before and after removing rows.

````python
# ==========================================
# 📏 Original Dataset Shape
# ==========================================

df.shape

# ==========================================
# 📏 Dataset Shape After dropna()
# ==========================================

clean_df.shape
````

# 🚀 Remove Columns with Missing Values

Instead of removing rows,

we can remove columns containing missing values.

### Syntax

```python
df.dropna(axis=1)
```

`axis=1`

means remove columns.

````python
# ==========================================
# 🗑️ Remove Columns Having Missing Values
# ==========================================

df.dropna(axis=1)
````

# 🚀 Remove Rows Only If All Values Are Missing

Sometimes a row contains only missing values.

In that case we can remove only those rows.

### Syntax

```python
df.dropna(how="all")
```

````python
# ==========================================
# 🗑️ Remove Rows Where Every Value is Missing
# ==========================================

df.dropna(how="all")
````

# 🚀 Remove Rows If Any Value is Missing

This is the default behavior.

### Syntax

```python
df.dropna(how="any")
```

````python
# ==========================================
# 🗑️ Remove Rows Having At Least One Missing Value
# ==========================================

df.dropna(how="any")
````

# 🚀 Remove Missing Values from Specific Columns

Sometimes we only care about important columns.

For example,

if the **Email** column is missing,

remove only those rows.

````python
# ==========================================
# 🗑️ Remove Rows Where Email is Missing
# ==========================================

df.dropna(subset=["email"])

# ==========================================
# 🗑️ Remove Rows Where Salary is Missing
# ==========================================

df.dropna(subset=["salary"])

# ==========================================
# 🗑️ Remove Rows Where Email and Salary are Missing
# ==========================================

df.dropna(subset=["email", "salary"])
````

# 🚀 Keep Rows Having Minimum Non-Missing Values

The `thresh` parameter keeps rows having at least a specified number of non-missing values.

### Syntax

```python
df.dropna(thresh=20)
```

If a row has fewer than 20 valid values,

it will be removed.

````python
# ==========================================
# 🗑️ Keep Rows with At Least 20 Non-Missing Values
# ==========================================

df.dropna(thresh=20)
````

# 🚀 Modify the Original Dataset

Normally,

`dropna()` returns a new DataFrame.

To permanently modify the original dataset,

use:

```python
inplace=True
```

⚠️ Use this carefully because it cannot be undone.

````python
# ==========================================
# 🗑️ Permanently Remove Missing Values
# ==========================================

temp_df = df.copy()

temp_df.dropna(inplace=True)

temp_df
````

# ✍️ Filling Missing Values using `fillna()`

---

# 🎯 Objective

Removing missing values is **not always the best solution**.

Sometimes deleting rows or columns can result in losing valuable information.

Instead of removing data, we can replace missing values with meaningful values using the **`fillna()`** function.

Some common techniques include:

✅ Fill with a constant value

✅ Fill with Mean

✅ Fill with Median

✅ Fill with Mode

✅ Forward Fill

✅ Backward Fill

> **Rule #3:** If the data is important, try filling missing values before deleting them.

````python
# ==========================================
# 📋 Check Missing Values
# ==========================================

df.isnull().sum()
````

# 🚀 Fill Missing Values with a Constant

Sometimes we simply want to replace every missing value with the same value.

### Syntax

```python
df.fillna(value)
```

````python
# ==========================================
# Replace Missing Values with 0
# ==========================================

df.fillna(0)

# ==========================================
# Replace Missing Values with "Unknown"
# ==========================================

df.fillna("Unknown")

# ==========================================
# Replace Missing Values with "Not Available"
# ==========================================

df.fillna("Not Available")
````

# 🚀 Fill Missing Values in a Specific Column

Instead of filling the entire dataset,

we can fill only one column.

````python
# ==========================================
# Fill Missing Salary with 0
# ==========================================

df["salary"] = df["salary"].fillna(0)

df.head()

# ==========================================
# Fill Missing Email with "Unknown"
# ==========================================

df["email"] = df["email"].fillna("Unknown")

df.head()
````

# 🚀 Fill Missing Values with Mean

The **Mean** is the average of all numerical values.

Best used for:

✔️ Continuous Numerical Data

Example:

- Salary
- Age
- Bonus

````python
# ==========================================
# Fill Missing Age with Mean
# ==========================================

temp_df = df.copy()

temp_df["age"] = pd.to_numeric(temp_df["age"], errors="coerce")

temp_df["age"] = temp_df["age"].fillna(temp_df["age"].mean())

temp_df.head()
````

# 🚀 Fill Missing Values with Median

The **Median** is the middle value after sorting.

It is useful when the dataset contains outliers.

Best used for:

✔️ Salary

✔️ Income

✔️ Property Price

````python
# ==========================================
# Fill Missing Salary with Median
# ==========================================

temp_df = df.copy()

temp_df["salary"] = pd.to_numeric(
    temp_df["salary"],
    errors="coerce"
)

temp_df["salary"] = temp_df["salary"].fillna(
    temp_df["salary"].median()
)

temp_df.head()
````

# 🚀 Fill Missing Values with Mode

The **Mode** is the most frequently occurring value.

Best used for:

✔️ Gender

✔️ City

✔️ Country

✔️ Department

````python
# ==========================================
# Fill Missing Department with Mode
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].fillna(
    temp_df["department"].mode()[0]
)

temp_df.head()
````

````python
# ==========================================
# Fill Missing Gender with Mode
# ==========================================

temp_df = df.copy()

temp_df["gender"] = temp_df["gender"].fillna(
    temp_df["gender"].mode()[0]
)

temp_df.head()
````

# 🚀 Fill Different Columns with Different Values

We can fill multiple columns at once by passing a dictionary.

````python
# ==========================================
# Fill Multiple Columns
# ==========================================

temp_df = df.copy()

temp_df.fillna({
    "salary": 0,
    "email": "Unknown",
    "department": "Not Assigned",
    "city": "Unknown"
})
````

# 🚀 Permanently Save Changes

The `inplace=True` parameter updates the original DataFrame directly.

⚠️ Use it carefully because the changes cannot be undone.

````python
# ==========================================
# Permanently Fill Missing Values
# ==========================================

temp_df = df.copy()

temp_df["department"].fillna(
    "Unknown",
    inplace=True
)

temp_df.head()
````

# 🎯 Summary

You have learned:

✅ `fillna()`

✅ Fill with Constant Value

✅ Fill with Mean

✅ Fill with Median

✅ Fill with Mode

✅ Fill Multiple Columns

✅ `inplace=True`

# ⏩ Forward Fill (`ffill`)

---

# 🎯 Objective

Forward Fill copies the **previous valid value** into the missing value.

It is mainly used when the previous observation is expected to remain valid.

Common Use Cases:

✅ Time Series Data

✅ Stock Prices

✅ Sensor Data

✅ Attendance Records

> **Rule #4:** Use Forward Fill only when the previous value logically belongs to the next missing record.

````python
# ==========================================
# 📋 Check Missing Values
# ==========================================

df.isnull().sum()
````

# 🚀 Forward Fill

### Syntax

```python
df.ffill()
```

or

```python
df.fillna(method="ffill")
```

````python
# ==========================================
# Forward Fill Entire Dataset
# ==========================================

df.ffill()
````

````python
# ==========================================
# Forward Fill a Single Column
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].ffill()

temp_df.head(15)
````

````python
# ==========================================
# Forward Fill Salary Column
# ==========================================

temp_df = df.copy()

temp_df["salary"] = temp_df["salary"].ffill()

temp_df.head(15)
````

# 🚀 Forward Fill with Limit

Sometimes we don't want to fill every missing value.

The **limit** parameter restricts how many consecutive missing values will be filled.

### Syntax

```python
df.ffill(limit=2)
```

````python
# ==========================================
# Forward Fill with Limit
# ==========================================

df.ffill(limit=2)
````

# ⏪ Backward Fill (`bfill`)

---

# 🎯 Objective

Backward Fill copies the **next valid value** into the missing value.

Instead of looking upward,

it looks downward for the next available value.

Common Use Cases:

✅ Future Schedule

✅ Timetable

✅ Forecast Data

✅ Sequential Records

````python
# ==========================================
# Backward Fill Entire Dataset
# ==========================================

df.bfill()
````

````python
# ==========================================
# Backward Fill Department
# ==========================================

temp_df = df.copy()

temp_df["department"] = temp_df["department"].bfill()

temp_df.head(15)
````

````python
# ==========================================
# Backward Fill Salary
# ==========================================

temp_df = df.copy()

temp_df["salary"] = temp_df["salary"].bfill()

temp_df.head(15)
````

# 🚀 Backward Fill with Limit

The **limit** parameter works the same as Forward Fill.

It limits the number of consecutive missing values that can be filled.

````python
# ==========================================
# Backward Fill with Limit
# ==========================================

df.bfill(limit=2)
````

# 📈 Interpolation

---

# 🎯 Objective

Interpolation estimates missing values using nearby numerical values.

Instead of copying values,

it calculates a reasonable value.

Common Use Cases:

✅ Temperature

✅ Rainfall

✅ Sales

✅ Stock Prices

✅ Scientific Data

> Interpolation should only be used on **numerical columns**.

````python
# ==========================================
# Check Numerical Columns
# ==========================================

df.select_dtypes(include="number").columns
````

````python
# ==========================================
# Interpolate Numerical Columns
# ==========================================

temp_df = df.copy()

numeric_columns = temp_df.select_dtypes(include="number").columns

temp_df[numeric_columns] = temp_df[numeric_columns].interpolate()

temp_df.head()
````

# 🚀 Different Interpolation Methods

Pandas supports multiple interpolation methods.

Some commonly used methods are:

- linear (Default)
- nearest
- quadratic
- cubic
- polynomial

The most commonly used method is **linear**.

````python
# ==========================================
# Linear Interpolation
# ==========================================

temp_df = df.copy()

numeric_columns = temp_df.select_dtypes(include="number").columns

temp_df[numeric_columns] = temp_df[numeric_columns].interpolate(
    method="linear"
)

temp_df.head()
````

````python
# ==========================================
# Nearest Interpolation
# ==========================================

temp_df = df.copy()

numeric_columns = temp_df.select_dtypes(include="number").columns

temp_df[numeric_columns] = temp_df[numeric_columns].interpolate(
    method="nearest"
)

temp_df.head()
````

# 📝 Best Practices

✅ Use **Mean** for normally distributed numerical data.

✅ Use **Median** when numerical columns contain outliers.

✅ Use **Mode** for categorical columns.

✅ Use **Forward Fill** for sequential or time-series data.

✅ Use **Backward Fill** when the next value is more appropriate.

✅ Use **Interpolation** only for numerical continuous data.

❌ Never replace every missing value with **0** without understanding the data.

❌ Never use Forward Fill or Backward Fill randomly.
